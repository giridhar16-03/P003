```python
def fit(x, y):
    m,n = x.shape
    theta = np.zeros(n)
    while True:
        theta_old = np.copy(theta)
        h_x = 1/(1 + np.exp(-x.dot(theta)))
        H = (x.T*h_x *(1-h_x)).dot(x) / m
        gradient_J_theta = x.T.dot(h_x - y) / m
        theta -= np.linalg.inv(H).dot(gradient_J_theta)
        if np.linalg.norm(theta-theta_old, ord=1) < 1e-5:
            break
    print(theta)
```

```python
fit(X_train, y_train)
```
